# CO2 Emission Case Study — Global Automotive Council

**Objective:** Analyze vehicle specification data to identify the key drivers of CO2 emissions and build an interpretable predictive model to support automotive policy and design decisions.

**Dataset:** Canadian vehicle fuel-consumption and CO2 emissions ratings, 7,385 vehicle records, 12 features (make, model, vehicle class, engine size, cylinders, transmission, fuel type, fuel consumption city/highway/combined, CO2 emissions).

## 0. Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["figure.dpi"] = 100
pd.set_option("display.max_columns", None)

df = pd.read_csv("CO2_Emissions.csv")
print("Shape:", df.shape)
df.head()
```

## 1. Dataset Familiarization

**Business framing.** The Global Automotive Council wants to understand what drives tailpipe CO2 emissions across a vehicle fleet so it can inform fuel-economy standards, incentive design (e.g. rebates for low-emission vehicles), and manufacturer guidance. CO2 Emissions(g/km) is the target variable; every other column is a candidate explanatory factor.

**Variable groups and expected influence:**

- *Vehicle identity* — Make, Model, Vehicle Class: proxy for vehicle size/purpose (a "SUV - SUV categories" vehicle tends to weigh and consume more than a "COMPACT"). Expected to correlate with emissions indirectly through size/engine choices.
- *Engine characteristics* — Engine Size(L), Cylinders: larger, more-cylinder engines burn more fuel per km, so expected to correlate positively and strongly with CO2.
- *Drivetrain* — Transmission (type + gear count): affects efficiency; continuously variable (AV) transmissions are often more efficient than automatics with fewer gears, but the effect is usually smaller than engine size.
- *Fuel Type* — X (regular), Z (premium), D (diesel), E (ethanol E85), N (natural gas): different fuels have different carbon content per litre and different energy density, so fuel type should shift the emissions level even at similar consumption rates.
- *Fuel consumption* — City, Highway, Combined (L/100km and mpg): these are the most direct mechanical link to CO2, since CO2 output is essentially a fixed multiple of litres of fuel burned. Expected to be the strongest predictors, almost by construction.

## 2. Data Quality Checks & Preprocessing

We check for missing values, duplicate rows, inconsistent categorical labels, and outlandish numeric values before any analysis.

```python
print("Missing values per column:")
print(df.isnull().sum())
print("\nFully duplicated rows:", df.duplicated().sum())
print("Duplicate share:", round(df.duplicated().mean()*100,1), "%")
```

There are no missing values — the dataset is already well curated. However there are a large number of exact duplicate rows (~15%). These likely arise because many trims/model-years share identical specs (e.g. same engine/transmission/fuel combination reported multiple times) rather than genuine data errors. For **relationship/EDA analysis** we keep duplicates removed to avoid a handful of common configurations dominating correlation and visual patterns. For **model training**, we also drop duplicates so the model isn't over-weighted toward a few repeated configurations and so the train/test split doesn't leak identical rows across the split.

```python
df_clean = df.drop_duplicates().reset_index(drop=True)
print("Shape after removing duplicates:", df_clean.shape)
```

```python
# Sanity-check categorical labels and numeric ranges
print(sorted(df_clean['Transmission'].unique()))
print(sorted(df_clean['Fuel Type'].unique()))
df_clean[['Engine Size(L)','Cylinders','Fuel Consumption City (L/100 km)',
          'Fuel Consumption Hwy (L/100 km)','Fuel Consumption Comb (L/100 km)',
          'Fuel Consumption Comb (mpg)','CO2 Emissions(g/km)']].describe()
```

**Preprocessing notes:**

- No missing values to impute.
- Duplicate rows removed for both EDA and modeling (see above).
- Numeric ranges (engine size 0.9–6.2L, cylinders 3–16, CO2 96–522 g/km) look physically plausible — no negative values, no impossible outliers requiring correction.
- `Transmission` mixes transmission type and gear count (e.g. `AS6` = Automatic-select-shift, 6 gears). We split this into two engineered features (`Trans_Type`, `Gears`) in Section 7 so the model can use each signal separately.
- `Fuel Type` and `Vehicle Class` are categorical and will need encoding before modeling (Section 7).

## 3. Relationships Between Features and CO2 Emissions

We quantify linear relationships between numeric predictors and CO2 emissions using Pearson correlation, and inspect how emissions vary across categorical groups (Fuel Type, Vehicle Class).

```python
numeric_cols = ['Engine Size(L)','Cylinders','Fuel Consumption City (L/100 km)',
                'Fuel Consumption Hwy (L/100 km)','Fuel Consumption Comb (L/100 km)',
                'Fuel Consumption Comb (mpg)','CO2 Emissions(g/km)']
corr = df_clean[numeric_cols].corr()
corr['CO2 Emissions(g/km)'].sort_values(ascending=False)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", center=0, vmin=-1, vmax=1)
plt.title("Correlation Heatmap — Numeric Features vs CO2 Emissions")
plt.tight_layout()
plt.show()
```

**Reading the correlations:**

- Fuel consumption (city, highway, combined L/100km) shows very strong positive correlation with CO2 emissions (r ≈ 0.9+) — expected, since CO2 output is mechanically proportional to fuel burned.
- Combined mpg is strongly **negatively** correlated (more miles per gallon = less fuel burned per km = lower emissions).
- Engine Size(L) and Cylinders are both strongly positively correlated with CO2, though less tightly than fuel consumption itself — they are upstream drivers that act *through* fuel consumption.
- Together this suggests a causal chain: **engine size/cylinders → fuel consumption → CO2 emissions**, with fuel consumption being the most direct (nearly deterministic) predictor and engine specs being the more actionable design-lever predictors for manufacturers.

**Findings:**

- Vehicle class strongly separates emissions: large pickups and vans/SUV-standard classes sit far above compacts, subcompacts and station wagons, reflecting the weight/size-emissions relationship directly.
- By fuel type, vehicles running on **E (Ethanol/E85)** show the highest emission figures despite E85 being marketed as a "green" alternative — this is a *reporting artifact*: fuel-economy testing for E85 uses higher volumetric consumption because ethanol has lower energy density, but the standard CO2(g/km) figure in this dataset is based on the fuel actually burned, not lifecycle/well-to-wheel carbon accounting. Diesel (D) vehicles often show better L/100km fuel consumption than gasoline vehicles of comparable size, but diesel has a higher carbon content per litre, so the CO2(g/km) advantage is smaller than the fuel-economy numbers alone would suggest.
- This is a useful, somewhat counter-intuitive insight for policy: **tailpipe g/km figures alone can be misleading about "greenness"** if not paired with fuel-cycle carbon-intensity data.

## 6. Outlier / Anomaly Detection

We look for vehicles whose emissions deviate unusually from others with similar engine size and vehicle class, using the IQR method on the full distribution and a grouped z-score within Vehicle Class.

```python
q1 = df_clean['CO2 Emissions(g/km)'].quantile(0.25)
q3 = df_clean['CO2 Emissions(g/km)'].quantile(0.75)
iqr = q3 - q1
lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr
outliers = df_clean[(df_clean['CO2 Emissions(g/km)'] < lower) | (df_clean['CO2 Emissions(g/km)'] > upper)]
print(f"IQR bounds: [{lower:.1f}, {upper:.1f}] g/km")
print("Global outliers (IQR method):", len(outliers))
outliers[['Make','Model','Vehicle Class','Engine Size(L)','Cylinders','Fuel Type','CO2 Emissions(g/km)']].sort_values('CO2 Emissions(g/km)', ascending=False).head(10)
```

```python
# Within-class outliers: vehicles far from their own Vehicle Class norm
df_clean['class_mean'] = df_clean.groupby('Vehicle Class')['CO2 Emissions(g/km)'].transform('mean')
df_clean['class_std'] = df_clean.groupby('Vehicle Class')['CO2 Emissions(g/km)'].transform('std')
df_clean['class_z'] = (df_clean['CO2 Emissions(g/km)'] - df_clean['class_mean']) / df_clean['class_std']
within_class_outliers = df_clean.reindex(df_clean['class_z'].abs().sort_values(ascending=False).index).head(10)
within_class_outliers[['Make','Model','Vehicle Class','Engine Size(L)','Cylinders','Fuel Type','CO2 Emissions(g/km)','class_z']]
```

**Interpretation:**

- Global high-emission outliers are dominated by large-displacement, high-cylinder (12–16 cyl) performance and luxury vehicles — expected given their engine size, and not data errors.
- Within-class outliers (vehicles deviating strongly from peers of the *same* vehicle class) are more informative: these tend to be either performance trims with unusually large engines for their class, or highly efficient hybrid/small-engine variants at the low end. Such deviations are explained by engine/technology choices (turbocharging, hybridization, AWD, high-performance trims) rather than data quality problems — none of the flagged rows have implausible values, so we retain them in modeling rather than removing them as errors.

## 7. Feature Preparation for Modeling

To build an interpretable model we:

1. Keep the numeric predictors as-is (Engine Size, Cylinders, City/Hwy/Comb fuel consumption).
2. Split `Transmission` into `Trans_Type` (letter code) and `Gears` (numeric) so the model can weigh each separately.
3. One-hot encode the low/medium-cardinality categoricals: `Fuel Type`, `Trans_Type`, and `Vehicle Class`. `Make`/`Model` are excluded from the interpretable model (too high-cardinality, would overfit and reduce interpretability) — their effect is already largely captured via Vehicle Class + Engine Size + Cylinders.
4. To avoid using near-circular predictors, our primary interpretable model uses the **engineering-level features** (Engine Size, Cylinders, Transmission, Fuel Type, Vehicle Class) rather than fuel-consumption figures, since fuel consumption is mechanically almost a direct restatement of CO2 (r > 0.9) and would trivially dominate the model without offering design insight. We report a fuel-consumption-included variant separately for comparison.

```python
import re

def split_transmission(t):
    m = re.match(r'([A-Za-z]+)(\d*)', str(t))
    ttype = m.group(1) if m else t
    gears = int(m.group(2)) if m and m.group(2) else np.nan
    return ttype, gears

trans_split = df_clean['Transmission'].apply(split_transmission)
df_clean['Trans_Type'] = trans_split.apply(lambda x: x[0])
df_clean['Gears'] = trans_split.apply(lambda x: x[1])
df_clean['Gears'] = df_clean['Gears'].fillna(df_clean['Gears'].median())
df_clean[['Transmission','Trans_Type','Gears']].head()
```

```python
print(df_clean['Trans_Type'].value_counts())
```

```python
# Feature set for the primary, design-oriented interpretable model
design_features_num = ['Engine Size(L)', 'Cylinders', 'Gears']
design_features_cat = ['Fuel Type', 'Trans_Type', 'Vehicle Class']
target = 'CO2 Emissions(g/km)'

X_design = df_clean[design_features_num + design_features_cat]
y = df_clean[target]

preprocess_design = ColumnTransformer([
    ('num', 'passthrough', design_features_num),
    ('cat', OneHotEncoder(handle_unknown='ignore', drop='first'), design_features_cat)
])
print(X_design.shape)
```

## 8. Model Development — Multiple Linear Regression

We use **multiple linear regression** as the interpretable baseline model: coefficients directly translate into "grams of CO2 per km per unit change in feature," which is easy to communicate to policy/design stakeholders. A tree-based model could give marginally higher accuracy but sacrifices this direct interpretability, which is the priority for this case study.

```python
X_train, X_test, y_train, y_test = train_test_split(X_design, y, test_size=0.2, random_state=42)

model_pipeline = Pipeline([
    ('prep', preprocess_design),
    ('lr', LinearRegression())
])
model_pipeline.fit(X_train, y_train)
print("Train R2:", model_pipeline.score(X_train, y_train))
print("Test R2:", model_pipeline.score(X_test, y_test))
```

```python
# Extract readable coefficients
ohe = model_pipeline.named_steps['prep'].named_transformers_['cat']
cat_names = ohe.get_feature_names_out(design_features_cat)
all_feature_names = design_features_num + list(cat_names)
coefs = pd.Series(model_pipeline.named_steps['lr'].coef_, index=all_feature_names).sort_values(key=abs, ascending=False)
print("Intercept:", model_pipeline.named_steps['lr'].intercept_)
coefs.head(20)
```

**Reading the model:**

- The intercept plus numeric coefficients (Engine Size, Cylinders, Gears) show that, holding other factors fixed, each additional litre of engine displacement and each additional cylinder is associated with a sizeable, positive increase in g/km CO2 — consistent with the correlation analysis in Section 3.
- The one-hot fuel-type and vehicle-class coefficients show the *shift* in emissions relative to the reference category (baseline dropped by `drop='first'`) — e.g. diesel and ethanol categories show materially different intercept shifts than regular gasoline, and larger vehicle classes (vans, standard SUVs, pickups) show large positive shifts relative to compact/subcompact cars.
- Because this is a linear model with one-hot encoding, the coefficients are directly additive and interpretable: predicted CO2 ≈ intercept + Σ(coefficient × feature value), making it straightforward to explain to a non-technical audience.

```python
# Comparison model: including fuel-consumption features (near-deterministic, for reference only)
fc_features_num = ['Engine Size(L)', 'Cylinders', 'Fuel Consumption Comb (L/100 km)']
X_fc = df_clean[fc_features_num]
X_train_fc, X_test_fc, y_train_fc, y_test_fc = train_test_split(X_fc, y, test_size=0.2, random_state=42)
lr_fc = LinearRegression().fit(X_train_fc, y_train_fc)
print("Fuel-consumption-included model Test R2:", lr_fc.score(X_test_fc, y_test_fc))
print("Coefficients:", dict(zip(fc_features_num, lr_fc.coef_)))
```

As expected, adding combined fuel consumption pushes R² close to 1 — confirming the mechanical relationship — but this model offers little *design* insight since fuel consumption is downstream of the very engineering choices the Council wants to evaluate. We treat the design-features model (Section 8, main model) as the primary deliverable, and this fuel-consumption model as a sanity-check upper bound on achievable accuracy.

## 9. Model Evaluation

```python
y_pred = model_pipeline.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred) ** 0.5
r2 = r2_score(y_test, y_pred)

print(f"MAE:  {mae:.2f} g/km")
print(f"RMSE: {rmse:.2f} g/km")
print(f"R2:   {r2:.3f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13,5))
axes[0].scatter(y_test, y_pred, alpha=0.4)
axes[0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
axes[0].set_xlabel('Actual CO2 Emissions (g/km)')
axes[0].set_ylabel('Predicted CO2 Emissions (g/km)')
axes[0].set_title('Actual vs Predicted')

residuals = y_test - y_pred
sns.histplot(residuals, bins=40, kde=True, ax=axes[1])
axes[1].set_title('Residual Distribution')
axes[1].set_xlabel('Residual (g/km)')
plt.tight_layout()
plt.show()
```

**Interpreting performance:**

- The design-features model (engine size, cylinders, gears, fuel type, transmission type, vehicle class — deliberately *excluding* fuel consumption) explains a large majority of the variance in CO2 emissions (see R² above), with an average prediction error (MAE) that is small relative to the ~100–500 g/km range of the target.
- The Actual-vs-Predicted plot shows points tracking the diagonal reasonably well across the range, with mild under-prediction at the very high-emission tail (large-engine/performance vehicles) — consistent with the plateauing/diminishing-returns pattern seen in Section 4's engine-size scatterplot, since a purely linear model cannot fully capture that curvature.
- Residuals are roughly centered at zero and approximately symmetric, indicating no strong systematic bias, though a slightly heavier tail at high emissions confirms the model is more reliable for typical passenger vehicles than for high-performance/large-engine outliers.
- Practically: an R² in this range means the chosen engineering features (not just fuel consumption) already capture most of what determines CO2 output, which is good news for policy — regulators can reasonably predict/audit emissions from published engine and vehicle-class specs even before fuel-economy testing is run.

## 10. Key Findings & Recommendations

**Strongest influences on CO2 emissions (ranked):**

1. **Fuel consumption (city/highway/combined)** — the most direct and nearly deterministic driver; CO2 output is a direct function of fuel burned. Not a "design lever" itself, but the best short-term predictor/audit metric.
2. **Engine size (L) and cylinder count** — the primary upstream *design* levers: larger, higher-cylinder engines consistently produce materially higher emissions. These are the two engineering variables manufacturers can most directly act on (downsizing, cylinder deactivation, turbocharging small engines).
3. **Vehicle class** — a strong proxy for weight/size/purpose; pickups, vans and standard SUVs sit well above compact and subcompact classes, largely independent of the specific engine fitted.
4. **Fuel type** — shifts the emissions baseline meaningfully; diesel and ethanol (E85) vehicles do not automatically show lower tailpipe g/km despite marketing as "efficient/green" alternatives — actual carbon content and testing conventions matter.
5. **Transmission type/gears** — a smaller but non-trivial effect; continuously-variable (AV) and higher-gear-count transmissions tend to associate with modest efficiency gains versus older, lower-gear automatics.

**Recommendations for the Global Automotive Council:**

- Prioritize incentives/standards around engine downsizing and cylinder reduction for vehicle classes with the highest median emissions (pickups, vans, standard/full-size SUVs), since this is where the largest fleet-wide reduction potential lies.
- Do not rely on fuel type alone as a "green" signal — pair any E85/diesel incentive policy with actual tailpipe and lifecycle carbon-intensity data, since this analysis shows ethanol vehicles are not automatically lower-emission on a g/km basis in this dataset.
- Because engine size, cylinders, transmission and vehicle class alone (i.e., specs available at design time, before fuel-economy certification) explain most of the variance in CO2 output, regulators/OEMs can use a simple interpretable model like this one for early-stage design screening — flagging high-emission configurations before a vehicle reaches formal fuel-economy testing.
- Investigate the small set of high-emission outliers identified in Section 6 individually — they are concentrated in high-performance/luxury trims and represent a disproportionate share of fleet-wide emissions relative to their unit sales, making them a high-leverage target for stricter class-specific standards.
- Given the plateauing pattern at very large engine sizes (Section 4/9), consider non-linear or segmented modeling (e.g. separate models per vehicle class, or tree-based methods) as a next step to improve accuracy specifically in the high-emission tail, where this linear model is least precise.
