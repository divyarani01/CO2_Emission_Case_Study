# 4. Data Cleaning

Based on Section 3, only two concrete issues were found: **duplicate rows** and **trailing
whitespace in `Model`**. We address exactly those, and nothing more — no feature engineering, no
encoding, no scaling. Every step below is verified with a before/after check.

## 4.1 Missing Values

Already confirmed as zero across all columns in Section 3.6. No action required.

```python
assert df.isnull().sum().sum() == 0, "Unexpected missing values found"
print("Confirmed: 0 missing values across all columns.")
```

## 4.2 Whitespace Issues

We strip leading/trailing whitespace from every text column. This is safe and non-destructive: it
only normalizes representation, it never changes the semantic value of a category.

```python
df_clean = df.copy()

for col in categorical_cols:
    df_clean[col] = df_clean[col].str.strip()

remaining_whitespace = sum(
    (df_clean[col].astype(str) != df_clean[col].astype(str).str.strip()).sum()
    for col in categorical_cols
)
print(f"Whitespace issues remaining after cleaning: {remaining_whitespace}")
```

## 4.3 Inconsistent Categorical Values

Section 3.8 found no case or spelling inconsistencies in `Make`, `Vehicle Class`, `Transmission`,
or `Fuel Type` — they were already clean. No mapping/standardization is required beyond the
whitespace strip above.

```python
for col in ["Make", "Vehicle Class", "Transmission", "Fuel Type"]:
    print(f"{col}: {df_clean[col].nunique()} unique, consistent categories")
```

## 4.4 Incorrect Datatypes

Numeric columns are already numeric. The only genuine improvement is storing the low-cardinality
categorical columns as pandas `category` dtype, which is more memory-efficient and communicates
intent clearly — this is a datatype correction, not feature engineering (no new information is
created, no encoding is applied).

```python
for col in ["Make", "Vehicle Class", "Transmission", "Fuel Type"]:
    df_clean[col] = df_clean[col].astype("category")

df_clean.dtypes
```

## 4.5 Invalid / Impossible Values

We re-verify that no numerical column contains zero, negative, or otherwise physically impossible
values (e.g. an engine size of 0L, or negative fuel consumption).

```python
for col in numerical_cols:
    n_invalid = (df_clean[col] <= 0).sum()
    print(f"{col:38s} -> min={df_clean[col].min():<8} invalid(<=0)={n_invalid}")
```

## 4.6 Duplicate Rows

This is the main cleaning action. Exact duplicate rows represent the same vehicle specification
repeated verbatim; keeping them would silently over-weight those specifications in every summary
statistic, plot, and future model. We drop them and keep the first occurrence.

```python
before = len(df_clean)
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
after = len(df_clean)

print(f"Rows before: {before:,}")
print(f"Rows after : {after:,}")
print(f"Removed    : {before - after:,} duplicate rows ({(before - after) / before:.1%})")
```

## 4.7 Cleaning Summary

| Step | Issue found | Action taken |
|---|---|---|
| Missing values | None | No action needed |
| Whitespace | 5 values in `Model` had trailing spaces | Stripped whitespace on all text columns |
| Inconsistent categories | None | No action needed |
| Incorrect datatypes | Categoricals stored as generic `object` | Converted to pandas `category` dtype |
| Invalid/impossible values | None | No action needed |
| Duplicate rows | ~1,100+ exact duplicates (~15% of data) | Dropped, keeping first occurrence |

From this point on, **all analysis uses `df_clean`**, the deduplicated and type-corrected dataset.
No rows are removed based on outlier status — that question is addressed on its own merits in
Section 5.F, since removing statistically-extreme-but-real vehicles would bias the dataset against
genuine high-performance/exotic cars.

```python
print(f"Final cleaned dataset: {df_clean.shape[0]:,} rows x {df_clean.shape[1]} columns")
df_clean.head()
```

# 5. Exploratory Data Analysis

With a clean, deduplicated dataset (`df_clean`), we now explore it in depth: first each variable on

```python
numerical_features = [
    "Engine Size(L)",
    "Cylinders",
    "Fuel Consumption City (L/100 km)",
    "Fuel Consumption Hwy (L/100 km)",
    "Fuel Consumption Comb (L/100 km)",
    "Fuel Consumption Comb (mpg)",
]
target_col = "CO2 Emissions(g/km)"
categorical_features = ["Make", "Vehicle Class", "Transmission", "Fuel Type"]

print("Numerical features :", numerical_features)
print("Target             :", target_col)
print("Categorical features:", categorical_features)
```

## A. Univariate Analysis

We look at each variable in isolation before considering relationships between variables. This
tells us the shape, spread, and skew of each numerical feature, and the dominant categories of each
categorical feature — the baseline understanding needed before any bivariate claim can be trusted.

### A.1 Numerical Features — Distribution, Skewness, Outliers

For each numerical predictor we plot a histogram+KDE and a boxplot side by side, using one subplot
grid instead of separate figures. Skewness is computed numerically alongside the visuals so we
don't have to eyeball asymmetry.

```python
def plot_distributions(data, columns, ncols=3):
    """Grid of histogram+KDE (top) and boxplot (bottom) per column."""
    nrows = -(-len(columns) // ncols)  # ceil division
    fig, axes = plt.subplots(nrows * 2, ncols, figsize=(6 * ncols, 4 * nrows * 2))
    axes = axes.reshape(nrows * 2, ncols)

    for i, col in enumerate(columns):
        r, c = divmod(i, ncols)
        hist_ax, box_ax = axes[2 * r, c], axes[2 * r + 1, c]

        sns.histplot(data[col], kde=True, ax=hist_ax, color="#3b6ea5", edgecolor="white")
        hist_ax.set_title(f"{col}\n(skew = {skew(data[col]):.2f})")
        hist_ax.set_xlabel(col)

        sns.boxplot(x=data[col], ax=box_ax, color="#f2a65a")
        box_ax.set_xlabel(col)

    # hide any unused axes
    for j in range(len(columns), nrows * ncols):
        r, c = divmod(j, ncols)
        axes[2 * r, c].axis("off")
        axes[2 * r + 1, c].axis("off")

    fig.suptitle("Univariate Distributions of Numerical Features", fontsize=16, y=1.0)
    fig.tight_layout()
    plt.show()


plot_distributions(df_clean, numerical_features)
```

```python
skew_table = df_clean[numerical_features].apply(skew).sort_values(ascending=False)
skew_table.to_frame("skewness")
```

**Interpretation:**

- Every numerical predictor is **moderately right-skewed** (skew roughly 0.8-1.1). None are
  symmetric, but none are extreme (skew > 2) either — this is typical for automotive specs, where
  most vehicles cluster around mainstream engine sizes/consumption and a smaller tail of larger,
  less efficient vehicles stretches the distribution rightward.
- `Cylinders` and `Fuel Consumption Hwy (L/100 km)` are the most skewed (~1.1) — a small number of
  high-cylinder, high-consumption vehicles (trucks, exotics) pull the tail out.
- The boxplots show a consistent pattern: a compact interquartile range with a cluster of high-end
  outliers above the upper whisker, and almost no low-end outliers. This matches the skew direction
  and foreshadows the Outlier Analysis in Section 5.F.
- Because none of these features are symmetric, models sensitive to feature distribution (e.g.
  linear regression with normality assumptions on residuals, or distance-based models) may benefit
  from a transformation later — a decision deferred to the feature engineering stage.

### A.2 Categorical Features — Frequency and Dominant Categories

For each categorical feature we show a countplot (ordered by frequency) and the exact percentage
distribution as a table. `Model` is excluded from this view: with over 2,000 unique values it
behaves as a near-identifier rather than a meaningful category for grouped analysis.

```python
def frequency_table(data, col, top_n=10):
    counts = data[col].value_counts()
    pct = (counts / len(data) * 100).round(2)
    return pd.DataFrame({"count": counts, "pct_%": pct}).head(top_n)


def plot_categorical_counts(data, columns, ncols=2, top_n=15):
    nrows = -(-len(columns) // ncols)
    fig, axes = plt.subplots(nrows, ncols, figsize=(8 * ncols, 5 * nrows))
    axes = axes.flatten()

    for ax, col in zip(axes, columns):
        top_categories = data[col].value_counts().head(top_n).index
        subset = data[data[col].isin(top_categories)]
        sns.countplot(
            data=subset, y=col, order=top_categories, ax=ax, color="#4c8577"
        )
        ax.set_title(f"{col} (top {min(top_n, data[col].nunique())} of {data[col].nunique()})")
        ax.set_xlabel("Count")

    for ax in axes[len(columns):]:
        ax.axis("off")

    fig.suptitle("Categorical Feature Frequencies", fontsize=16, y=1.02)
    fig.tight_layout()
    plt.show()


plot_categorical_counts(df_clean, categorical_features)
```

```python
for col in categorical_features:
    print(f"--- {col} ---")
    display(frequency_table(df_clean, col))
    print()
```

**Interpretation:**

- **Fuel Type** is heavily concentrated in two categories: `X` (regular gasoline) and `Z` (premium
  gasoline) together account for the large majority of vehicles; `E` (ethanol/E85) and `D` (diesel)
  are minority categories, and `N` (natural gas) is a single-vehicle category — any conclusion drawn
  about natural-gas vehicles specifically will rest on effectively no sample size and should be
  treated as anecdotal, not statistical.
- **Vehicle Class** is dominated by mainstream passenger segments (`SUV - SMALL`, `MID-SIZE`,
  `COMPACT`), with commercial/specialty classes (`VAN - CARGO`, `PICKUP TRUCK`) much rarer.
- **Transmission** shows automatic variants (`AS6`, `AS8`, `A6`, etc.) as dominant, consistent with
  the broader market; fully manual transmissions are a small minority.
- **Make** has 42 categories, several of them low-volume luxury/exotic brands. This long tail is
  worth keeping in mind for Section 5.C, where small-sample categories can produce visually striking
  but statistically fragile average-CO2 comparisons.

## B. Target Variable Analysis — CO2 Emissions(g/km)

Before relating the target to any predictor, we characterize it on its own: its shape, spread, and
whether it contains extreme values. Every regression modeling decision downstream (e.g. whether to
transform the target, which metric to prioritize) depends on this.

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 5))

sns.histplot(df_clean[target_col], kde=True, ax=axes[0], color="#c0392b", edgecolor="white")
axes[0].set_title("Histogram + KDE")
axes[0].set_xlabel(target_col)

sns.kdeplot(df_clean[target_col], fill=True, ax=axes[1], color="#c0392b")
axes[1].set_title("KDE (density)")
axes[1].set_xlabel(target_col)

sns.boxplot(x=df_clean[target_col], ax=axes[2], color="#e67e22")
axes[2].set_title("Boxplot")
axes[2].set_xlabel(target_col)

fig.suptitle("Distribution of Target Variable: CO2 Emissions(g/km)", fontsize=16, y=1.03)
fig.tight_layout()
plt.show()
```

```python
target_skew = skew(df_clean[target_col])
q1, q3 = df_clean[target_col].quantile([0.25, 0.75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
n_outliers = ((df_clean[target_col] < lower) | (df_clean[target_col] > upper)).sum()

print(f"Skewness        : {target_skew:.3f}")
print(f"IQR bounds      : [{lower:.1f}, {upper:.1f}]")
print(f"Outlier count   : {n_outliers} ({n_outliers / len(df_clean):.2%})")
df_clean[target_col].describe()
```

**Interpretation:** `CO2 Emissions(g/km)` ranges from 96 to 522 g/km with a mean around 251 g/km.
Its skewness (~0.56) is the mildest of all the numerical columns — noticeably more symmetric than
the predictors, though still right-tailed. The boxplot shows a modest cluster of high-emission
outliers (roughly 1% of records) rather than a heavy tail, consistent with a small number of
high-performance/large vehicles. Because the skew is mild, the target likely does **not** need a
log or Box-Cox transformation for a linear model — a decision to revisit empirically at the
modeling stage, but the EDA evidence here does not demand one.

## C. Bivariate Analysis

Now we relate each predictor to the target. Numerical predictors are examined with scatterplots and
a fitted regression line (to visually gauge linearity and strength); categorical predictors are
examined with boxplots/violin plots per category, plus the average target value per category.

### C.1 Numerical Features vs Target

```python
def plot_scatter_with_regression(data, columns, target, ncols=3):
    nrows = -(-len(columns) // ncols)
    fig, axes = plt.subplots(nrows, ncols, figsize=(6.5 * ncols, 5 * nrows))
    axes = axes.flatten()

    for ax, col in zip(axes, columns):
        sns.regplot(
            data=data, x=col, y=target, ax=ax,
            scatter_kws={"alpha": 0.25, "s": 15, "color": "#3b6ea5"},
            line_kws={"color": "#c0392b", "linewidth": 2},
        )
        r = data[col].corr(data[target])
        ax.set_title(f"{col}\n(r = {r:.2f})")

    for ax in axes[len(columns):]:
        ax.axis("off")

    fig.suptitle(f"Numerical Features vs {target}", fontsize=16, y=1.02)
    fig.tight_layout()
    plt.show()


plot_scatter_with_regression(df_clean, numerical_features, target_col)
```

```python
df_clean[numerical_features + [target_col]].corr()[target_col].drop(target_col).sort_values(
    key=abs, ascending=False
).to_frame("correlation_with_target")
```

**Interpretation:**

- **`Fuel Consumption City (L/100 km)` (r ~ 0.92)** and **`Fuel Consumption Comb (L/100 km)`
  (r ~ 0.92)** show the strongest, most linear relationships with CO2 emissions — unsurprising,
  since fuel burned is almost mechanically proportional to CO2 produced.
- **`Fuel Consumption Comb (mpg)` (r ~ -0.91)** shows an equally strong but *curved* negative
  relationship — because mpg is the inverse of a consumption rate, its relationship with a linearly
  related target is naturally non-linear even though the correlation coefficient is high. This is
  an early hint that mpg contributes no new information beyond the other consumption columns
  (confirmed formally in Section 5.E).
- **`Engine Size(L)` (r ~ 0.86)** and **`Cylinders` (r ~ 0.83)** are strong but visibly noisier —
  many points are scattered further from the regression line, since engine size/cylinder count
  determine emissions only indirectly, through fuel consumption.
- All six numerical features are strongly, positively (or for mpg, inversely) related to CO2
  emissions — there is no weak or negligible numerical predictor in this dataset.

### C.2 Categorical Features vs Target

```python
def plot_categorical_vs_target(data, columns, target, top_n=15):
    fig, axes = plt.subplots(2, len(columns), figsize=(7 * len(columns), 12))

    for i, col in enumerate(columns):
        order = data.groupby(col, observed=True)[target].median().sort_values(ascending=False).index
        if len(order) > top_n:
            order = order[:top_n]
        subset = data[data[col].isin(order)]

        sns.boxplot(data=subset, x=col, y=target, order=order, ax=axes[0, i], color="#4c8577")
        axes[0, i].set_title(f"{col} vs {target} (Boxplot)")
        axes[0, i].tick_params(axis="x", rotation=90)

        sns.violinplot(data=subset, x=col, y=target, order=order, ax=axes[1, i], color="#8e7cc3")
        axes[1, i].set_title(f"{col} vs {target} (Violin)")
        axes[1, i].tick_params(axis="x", rotation=90)

    fig.suptitle(f"Categorical Features vs {target}", fontsize=18, y=1.01)
    fig.tight_layout()
    plt.show()


plot_categorical_vs_target(df_clean, categorical_features, target_col)
```

```python
for col in categorical_features:
    print(f"--- Average {target_col} by {col} (top 8) ---")
    display(
        df_clean.groupby(col, observed=True)[target_col]
        .agg(["mean", "count"])
        .sort_values("mean", ascending=False)
        .round(1)
        .head(8)
    )
    print()
```

**Interpretation:**

- **Fuel Type:** Ethanol (`E`) and premium gasoline (`Z`) vehicles show the *highest* average CO2,
  regular gasoline (`X`) and diesel (`D`) sit lower, and natural gas (`N`) shows the lowest — but
  `N` has exactly one vehicle in the data, so that ranking is not statistically meaningful. The
  more reliable read is that `E`/`Z` vehicles average higher CO2 than `X`/`D`, largely because
  ethanol-flex and premium-fuel vehicles in this dataset skew toward larger, more powerful engines
  — this is confounded with engine size, not a pure "fuel chemistry" effect.
- **Vehicle Class:** `VAN - PASSENGER` and `VAN - CARGO` have the highest average CO2 (~360-400
  g/km), while `STATION WAGON - SMALL` and `COMPACT` have the lowest (~200-220 g/km). This tracks
  intuitively with vehicle size/weight and engine requirements.
- **Transmission:** Higher-gear automatic transmissions (`A7`, `A10`) associate with the highest
  average CO2, though sample sizes per transmission code vary and some categories are thin.
- **Make:** Exotic/performance brands (Bugatti, Lamborghini, Rolls-Royce, Bentley) sit far above
  the rest — several of these are single- or few-vehicle categories, so they read as legitimate
  high-performance outliers rather than representative brand-level effects (see Section 5.F).

## D. Correlation Analysis

We now look at all numerical variables together (predictors + target) via a single correlation
matrix and heatmap, to see the full web of relationships at once, not just each predictor's
relationship with the target in isolation.

```python
corr_matrix = df_clean[numerical_features + [target_col]].corr()

fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(
    corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0,
    linewidths=0.5, square=True, ax=ax,
)
ax.set_title("Correlation Heatmap — Numerical Features & Target", fontsize=14, pad=12)
fig.tight_layout()
plt.show()
```

**Interpretation:**

- **Strong positive correlations:** All fuel-consumption measures (`City`, `Hwy`, `Comb L/100km`)
  correlate with each other at **r > 0.85**, and each correlates strongly with the target
  (r ~ 0.88-0.92). `Engine Size(L)` and `Cylinders` also correlate strongly with each other
  (r ~ 0.93) and with the target (r ~ 0.83-0.86).
- **Strong negative correlation:** `Fuel Consumption Comb (mpg)` correlates negatively with
  everything else (r as low as -0.91 with the target, and around -0.87 to -0.93 with the other
  consumption columns), because mpg is an inverse-scaled version of the same underlying quantity.
- **Weak correlations:** there are essentially none among these six numerical columns — every pair
  shows at least a moderate-to-strong relationship. This dataset does not contain a "quiet"
  numerical feature that is only weakly tied to the rest.
- **Why the fuel consumption variables are so highly correlated:** `Fuel Consumption Comb (L/100 km)`
  is, by definition, a weighted average of `City` and `Hwy` (55% city + 45% highway, per the
  dataset documentation) — it is mathematically derived from the other two, not an independent
  measurement. `Fuel Consumption Comb (mpg)` is, in turn, a unit conversion of the combined L/100km
  figure (`mpg ~ 235.2 / (L/100km)`), i.e. its mathematical inverse rescaled to a different unit
  system. All four consumption-related columns are therefore capturing **one underlying signal**
  (how much fuel the vehicle burns) expressed four different ways. This is the exact pattern that
  produces severe multicollinearity, examined formally next.

## E. Multicollinearity Analysis

Correlation tells us about *pairwise* linear relationships. Multicollinearity is a broader problem:
a feature can be near-perfectly predictable from a *combination* of several other features even if
no single pairwise correlation looks extreme. The Variance Inflation Factor (VIF) quantifies this.

**What VIF means:** for each predictor, VIF regresses that predictor on all the *other* predictors
and measures how well they explain it (R-squared of that auxiliary regression). Formally,
`VIF = 1 / (1 - R^2)`. A VIF of 1 means no multicollinearity; conventional thresholds flag VIF > 5
or VIF > 10 as concerning, since it means a large share of that feature's variance is redundant with
other features already in the model.

```python
X_vif = df_clean[numerical_features]

vif_table = pd.DataFrame({
    "feature": X_vif.columns,
    "VIF": [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])],
}).sort_values("VIF", ascending=False).reset_index(drop=True)

vif_table
```

**Interpretation:**

- `Fuel Consumption Comb (L/100 km)`, `Fuel Consumption City (L/100 km)`, and
  `Fuel Consumption Hwy (L/100 km)` all show **extremely high VIF values (in the thousands to tens
  of thousands)** — this is not a subtle statistical nuance, it is a mathematical identity: combined
  consumption is a fixed linear formula of city and highway consumption (`0.55 x City + 0.45 x Hwy`),
  so including all three together means one column is an exact linear function of the other two.
  VIF explodes toward infinity in exactly this situation.
- `Engine Size(L)` and `Cylinders` show **moderate-to-high VIF (roughly 55-80)** — still well above
  the conventional VIF > 10 danger threshold, driven by the fact that larger engines mechanically
  require more cylinders, so the two measure overlapping physical characteristics of the powertrain.
- `Fuel Consumption Comb (mpg)` shows the **lowest VIF (~7)** among the six — still above the
  conservative threshold of 5, because it remains a rescaled inverse of the combined L/100km figure,
  but its non-linear (reciprocal) relationship to the others means the *linear* auxiliary regression
  used by VIF cannot fit it as perfectly as it fits the three L/100km-based columns.
- **Mathematically related groups identified:**
  - `{City, Hwy, Comb (L/100 km)}` — a closed linear system (combined is a weighted sum of the
    other two); at most one of these three should ever appear together in a linear model.
  - `Comb (mpg)` — a nonlinear (reciprocal) transform of `Comb (L/100 km)`; carries essentially the
    same information as `Comb (L/100 km)` in a different unit and shape.
  - `{Engine Size(L), Cylinders}` — not mathematically identical, but physically and statistically
    entangled powertrain-size measures.
- **Practical implication (no features removed here, per scope):** a linear regression using all six
  numerical columns simultaneously would produce unstable, unreliable coefficients — the model
  could not tell which of the four fuel-consumption-flavored columns "deserves credit" for
  explaining CO2, and small data changes could flip coefficient signs. When feature selection is
  performed later, at most one fuel-consumption column (and a considered choice between engine size
  and cylinders, or a combination via regularization) should be retained for a linear model. Tree-
  based models are far less sensitive to this issue and could reasonably use all six as-is.

## F. Outlier Analysis

We now quantify outliers formally using the IQR method (values beyond 1.5x the interquartile range
from Q1/Q3), across all numerical columns including the target.

```python
def iqr_outlier_summary(data, columns):
    rows = []
    for col in columns:
        q1, q3 = data[col].quantile([0.25, 0.75])
        iqr = q3 - q1
        lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
        n_out = ((data[col] < lower) | (data[col] > upper)).sum()
        rows.append({
            "feature": col, "lower_bound": round(lower, 2), "upper_bound": round(upper, 2),
            "n_outliers": n_out, "pct_outliers": round(100 * n_out / len(data), 2),
        })
    return pd.DataFrame(rows).sort_values("n_outliers", ascending=False).reset_index(drop=True)


outlier_summary = iqr_outlier_summary(df_clean, numerical_features + [target_col])
outlier_summary
```

```python
all_num_cols = numerical_features + [target_col]
fig, axes = plt.subplots(1, len(all_num_cols), figsize=(3.2 * len(all_num_cols), 6))

for ax, col in zip(axes, all_num_cols):
    sns.boxplot(y=df_clean[col], ax=ax, color="#e67e22")
    ax.set_title(col, fontsize=10)
    ax.set_ylabel("")

fig.suptitle("Boxplots — Outlier Detection Across All Numerical Columns", fontsize=15, y=1.03)
fig.tight_layout()
plt.show()
```

**Interpretation:** Every numerical column shows a modest share of IQR-flagged outliers, roughly
1.2%-2.8% of records, concentrated entirely on the high end (large engines, high consumption, high
cylinder counts, high emissions) with virtually no low-end outliers. `Cylinders` and
`Fuel Consumption Hwy (L/100 km)` show the largest outlier shares, consistent with their higher
skew from Section 5.A.

**Should these outliers be removed?** No, not at this stage, and likely not at all for this
business problem:

- The values are not data-entry errors — they correspond to genuine vehicle categories (large
  trucks, SUVs, and high-performance/exotic cars) with legitimately higher engine sizes, fuel
  consumption, and emissions. Removing them would not "clean" the data; it would **delete real,
  policy-relevant vehicles** from a dataset whose entire purpose is understanding emissions across
  the full spectrum of vehicles the Global Automotive Council regulates.
- The outlier rate is low (under 3% per column) and concentrated in the same direction as the
  underlying skew, not scattered randomly — this is the signature of a heavy-tailed but genuine
  distribution, not contamination.
- The recommended path is to **retain all outliers through EDA** and revisit the question only at
  modeling time, where techniques like robust scaling, regularization, or tree-based models (which
  are naturally robust to outliers) can handle them without discarding data. If a linear model later
  proves highly sensitive to these points, capping (winsorizing) is a safer alternative to outright
  removal, since it preserves the ordering and presence of extreme vehicles.

### 3. Potential Multicollinearity

- `Fuel Consumption City`, `Hwy`, and `Comb (L/100 km)` are mathematically related (combined is a

### 4. Interesting Business Insights

- Vehicle *class* and *size* (vans, SUVs, trucks) separate emissions more dramatically than fuel
  type alone — suggesting fleet composition (what kinds of vehicles are sold/driven) may be as
  important a policy lever as fuel-type mandates.
- The apparent "fuel type" effect on CO2 (e.g. premium gasoline vehicles averaging higher emissions

- A small number of ultra-low-volume, high-emission luxury/performance brands (Bugatti, Lamborghini,
  Rolls-Royce, Bentley) pull "by-Make" averages upward disproportionately; brand-level comparisons
  should be read alongside sample size.

### 5. Variables Expected to Be Highly Predictive of CO2 Emissions

Based on correlation strength and VIF-aware reasoning, a linear model should expect the strongest
signal from **one** representative fuel-consumption measure (e.g. `Fuel Consumption Comb (L/100 km)`,
which had the highest correlation with the target after `City`) plus `Vehicle Class`, with
`Engine Size(L)`/`Cylinders` and `Fuel Type` contributing secondary, partially redundant signal. A
non-linear/tree-based model could reasonably use all numerical features together, since it is not
sensitive to the multicollinearity identified above.

### 6. Recommendations Before Feature Engineering and Model Building

- Carry `df_clean` forward — do not re-introduce the dropped duplicates.
- For linear models: select at most one fuel-consumption column and be deliberate about whether to
  include both `Engine Size(L)` and `Cylinders`, or consider regularization (Ridge/Lasso) that can
  tolerate correlated predictors.
- For tree-based/ensemble models: multicollinearity is much less of a concern, so all numerical
  features can be retained; VIF-driven feature removal is primarily a linear-model concern.
- Do not remove outliers outright; consider capping only if a chosen linear model proves unstable.
- When encoding categorical features later, treat `Make` and `Transmission` as higher-cardinality
  fields that may benefit from grouping rare categories, and exclude `Model` from modeling entirely
  given its near-identifier cardinality (2,048 unique values).

## Task 1 — Transmission Feature Engineering

`Transmission` currently packs two pieces of information into one string: a **family** (the
mechanism — Automatic, Automated Manual, Automatic Select Shift, Continuously Variable, or Manual)
and a **gear count** (a number suffix, where applicable). Before deciding whether to keep, discard,
or split the gear-count information, we test empirically whether it carries meaningful signal about
CO2 emissions, rather than assuming either way.

```python
transmission_family = df_clean["Transmission"].str.extract(r"([A-Za-z]+)")[0]
transmission_gears = df_clean["Transmission"].str.extract(r"(\d+)")[0].astype(float)

print("Transmission family counts:")
print(transmission_family.value_counts())
```

### 1.1 CO2 Emissions Within Each Transmission Family

For each family, we compare the CO2 distribution across its gear-count variants side by side.

```python
def plot_family_gear_boxplots(data, family_labels, target, families):
    fig, axes = plt.subplots(1, len(families), figsize=(6.5 * len(families), 5), sharey=True)

    for ax, fam in zip(axes, families):
        subset = data.loc[family_labels == fam, ["Transmission", target]].copy()
        order = sorted(subset["Transmission"].unique(), key=lambda x: (len(x), x))
        sns.boxplot(data=subset, x="Transmission", y=target, order=order, ax=ax, color="#4c8577")
        ax.set_title(f"Family: {fam}")
        ax.tick_params(axis="x", rotation=45)
        ax.set_xlabel("")

    fig.suptitle("CO2 Emissions by Gear-Count Variant, Within Each Transmission Family",
                 fontsize=15, y=1.03)
    fig.tight_layout()
    plt.show()


plot_family_gear_boxplots(
    df_clean, transmission_family, target_col, families=["A", "AS", "AM", "M"]
)
```

```python
def gear_variant_summary(data, family_labels, target, families):
    frames = []
    for fam in families:
        subset = data.loc[family_labels == fam]
        summary = (
            subset.groupby("Transmission", observed=True)[target]
            .agg(["mean", "count"])
            .round(1)
        )
        summary.insert(0, "family", fam)
        frames.append(summary)
    return pd.concat(frames)


gear_variant_summary(df_clean, transmission_family, target_col, families=["A", "AS", "AM", "M"])
```

### 1.2 Does Gear Count Add Meaningful Predictive Information?

We quantify this directly: the correlation between gear count and CO2 emissions, computed
**within** each family (so family-level differences don't confound the comparison), plus the
family-level averages for context.

```python
family_means = (
    df_clean.groupby(transmission_family, observed=True)[target_col]
    .agg(["mean", "count"])
    .sort_values("mean", ascending=False)
)
print("Family-level average CO2 emissions:")
print(family_means.round(1))
print()

print("Correlation between gear count and CO2, computed WITHIN each family:")
for fam in ["A", "AS", "AM"]:
    mask = transmission_family == fam
    r = transmission_gears[mask].corr(df_clean.loc[mask, target_col])
    print(f"  {fam:3s}: r = {r:+.3f}  (n={mask.sum()})")
```

**Interpretation:**

- **Family-level differences are large and consistent:** average CO2 ranges from ~184 g/km (`AV`,
  continuously variable) up to ~278 g/km (`A`, conventional automatic) — a ~95 g/km spread that is
  substantial relative to the target's own range (96-522 g/km).
- **Within-family gear-count correlations are weak and inconsistent in sign:** r = -0.15 within
  family `A`, but r = +0.20 within `AS` and r = +0.37 within `AM`. If gear count were a genuine,
  generalizable driver of emissions, we would expect the same direction of relationship across
  families; instead the sign flips, meaning whatever weak association exists is specific to a
  particular family's model/engine mix rather than a stable "more gears -> more/less CO2" effect.
- **Domain knowledge reinforces this:** `AV` (continuously variable) transmissions have no discrete
  gear count at all by design, so a gear-count feature is not even well-defined for roughly 7% of
  the fleet — any encoding of gear count would require an arbitrary placeholder for these vehicles.
- **Conclusion: gear count does not contribute meaningful, generalizable predictive information.**
  The transmission *family* is the informative part of this column; the gear-count suffix is noisy,
  inconsistent, and undefined for one whole family. We extract the family only.

```python
df_clean["Transmission_Family"] = transmission_family.astype("category")
df_clean = df_clean.drop(columns=["Transmission"])

print("New column value counts:")
print(df_clean["Transmission_Family"].value_counts())
```

**Conclusion (Task 1):** `Transmission` is replaced by `Transmission_Family` (5 categories: `A`,
`AS`, `AM`, `AV`, `M`), discarding the gear-count suffix. This is appropriate because the gear count
showed no consistent, family-independent relationship with CO2 emissions, is undefined for
continuously-variable transmissions, and — most importantly for a linear model — dropping it avoids
inflating the categorical feature space with dozens of low-frequency, low-signal transmission codes.

## Task 2 — Engine Feature Engineering

EDA found `Engine Size(L)` and `Cylinders` strongly correlated with each other (r ~ 0.93) and both
strongly correlated with the target — the classic multicollinearity setup. Rather than blindly
dropping one, we construct an engineered alternative, `Engine_Size_Per_Cylinder` (displacement per
cylinder — a proxy for whether the engine is a small number of large cylinders vs. many small ones,
e.g. turbocharged-downsized designs vs. large naturally-aspirated engines), and compare all three
representations on equal footing.

```python
df_clean["Engine_Size_Per_Cylinder"] = df_clean["Engine Size(L)"] / df_clean["Cylinders"]

engine_candidates = ["Engine Size(L)", "Cylinders", "Engine_Size_Per_Cylinder"]
df_clean[engine_candidates].describe().T
```

### 2.1 Correlation With Target

```python
engine_target_corr = (
    df_clean[engine_candidates + [target_col]]
    .corr()[target_col]
    .drop(target_col)
    .sort_values(key=abs, ascending=False)
)
engine_target_corr.to_frame("correlation_with_target")
```

### 2.2 Variance Inflation Factor — All Three Together, and Paired Alternatives

```python
def compute_vif(data, columns):
    X = data[columns]
    return pd.DataFrame({
        "feature": columns,
        "VIF": [variance_inflation_factor(X.values, i) for i in range(X.shape[1])],
    }).sort_values("VIF", ascending=False).reset_index(drop=True)


print("VIF — all three engine representations together:")
display(compute_vif(df_clean, engine_candidates))

print("\nVIF — Engine Size(L) + Engine_Size_Per_Cylinder only:")
display(compute_vif(df_clean, ["Engine Size(L)", "Engine_Size_Per_Cylinder"]))

print("\nVIF — Cylinders + Engine_Size_Per_Cylinder only:")
display(compute_vif(df_clean, ["Cylinders", "Engine_Size_Per_Cylinder"]))
```

**Interpretation:**

- `Engine Size(L)` has the strongest standalone correlation with the target (r ~ 0.855), narrowly
  ahead of `Cylinders` (r ~ 0.835); `Engine_Size_Per_Cylinder` is markedly weaker on its own
  (r ~ 0.589) — the ratio trades away most of the raw-magnitude signal that makes engine size/
  cylinder count predictive in the first place.
- With all three together, VIF is high for `Engine Size(L)` (~40) and `Cylinders` (~46) — confirming
  they are largely redundant with each other, exactly as EDA suggested.
- Pairing `Engine Size(L)` with `Engine_Size_Per_Cylinder` still leaves VIF around **11 for both** —
  above the conventional VIF > 10 concern threshold — because the ratio is still partly explainable
  by engine size (they share the numerator). Pairing `Cylinders` with the ratio behaves similarly
  (VIF ~13).
- **Business interpretation:** `Engine Size(L)` is also the more standard, directly interpretable
  automotive specification (raw displacement) used in real emissions-policy discussions (e.g. engine
  downsizing mandates), whereas the per-cylinder ratio is a more abstract engineering-design signal
  that adds complexity without a corresponding gain in predictive strength here.

**Conclusion (Task 2):** keep **`Engine Size(L)` only**. It has the highest standalone correlation
with the target among the three candidates, is the most interpretable for business/policy framing,
and dropping both `Cylinders` and `Engine_Size_Per_Cylinder` removes the most severe multicollinearity
(VIF ~40-46) at a small cost, since the ratio's own standalone predictive power is materially weaker
and still leaves elevated VIF even when combined with engine size. `Engine_Size_Per_Cylinder` is left
in `df_clean` for reference/analysis but is **not** carried into the modeling feature set.

## Task 3 — Fuel Consumption Feature Engineering

Four columns describe fuel consumption: `City`, `Hwy`, `Comb (L/100 km)`, and `Comb (mpg)`. EDA
already flagged extreme VIF among these. Here we formalize the decision of which, if any, to retain.

```python
fuel_candidates = [
    "Fuel Consumption City (L/100 km)",
    "Fuel Consumption Hwy (L/100 km)",
    "Fuel Consumption Comb (L/100 km)",
    "Fuel Consumption Comb (mpg)",
]

fig, axes = plt.subplots(1, 3, figsize=(19, 5.5))
pairs = [
    ("Fuel Consumption City (L/100 km)", "Fuel Consumption Comb (L/100 km)"),
    ("Fuel Consumption Hwy (L/100 km)", "Fuel Consumption Comb (L/100 km)"),
    ("Fuel Consumption Comb (L/100 km)", "Fuel Consumption Comb (mpg)"),
]
for ax, (x_col, y_col) in zip(axes, pairs):
    sns.scatterplot(data=df_clean, x=x_col, y=y_col, ax=ax, alpha=0.25, s=15, color="#3b6ea5")
    r = df_clean[x_col].corr(df_clean[y_col])
    ax.set_title(f"r = {r:.3f}")

fig.suptitle("Pairwise Relationships Among Fuel Consumption Measures", fontsize=15, y=1.03)
fig.tight_layout()
plt.show()
```

```python
print("Correlation matrix (fuel consumption measures + target):")
display(df_clean[fuel_candidates + [target_col]].corr().round(3))

print("\nVIF — all four fuel consumption measures together:")
display(compute_vif(df_clean, fuel_candidates))

print("\nCorrelation with target, ranked:")
fuel_target_corr = (
    df_clean[fuel_candidates + [target_col]]
    .corr()[target_col]
    .drop(target_col)
    .sort_values(key=abs, ascending=False)
)
display(fuel_target_corr.to_frame("correlation_with_target"))
```

**Interpretation:**

- All four measures correlate with the target almost equally strongly (|r| between 0.88 and 0.92);
  `City` edges out `Comb (L/100 km)` by less than 0.002 in raw correlation — a difference with no
  practical significance, not a basis for choosing between them on statistics alone.
- VIF for all four together runs into the **tens of thousands** — a mathematical certainty, not a
  borderline judgment call: `Fuel Consumption Comb (L/100 km)` correlates with `City` at r = 0.994
  and with `Hwy` at r = 0.951, because it is *defined* as their weighted average
  (0.55 x City + 0.45 x Hwy, per the dataset documentation). `Comb (mpg)` correlates at r = -0.928
  with `Comb (L/100 km)` because it is a reciprocal unit conversion of the same quantity
  (mpg ~ 235.2 / (L/100km)).
- **Statistical + domain justification for keeping `Fuel Consumption Comb (L/100 km)` only:** since
  it is mathematically constructed from both `City` and `Hwy`, it already captures effectively all
  of the information in those two columns (r = 0.994 and 0.951 respectively) in a single number,
  representing real-world blended driving conditions — exactly the quantity regulators and
  consumers care about. `Comb (mpg)` is dropped as a pure redundant rescaling of the same figure.
  Choosing `City` instead would discard genuine highway-driving information that `Comb` already
  folds in, for a correlation gain of essentially zero.

**Conclusion (Task 3):** keep **`Fuel Consumption Comb (L/100 km)` only**; drop `City`, `Hwy`, and
`Comb (mpg)` as redundant, both mathematically and statistically.

# Stage 6: Feature Selection

With the engineering decisions from Stage 5 in hand, we assemble the final feature set. The goal
here is **not** dimensionality reduction for its own sake — it is removing features that are
mathematically or near-mathematically redundant with another retained feature, while keeping every
feature that carries distinct information, using only correlation, VIF, and domain reasoning (no
model-based selection methods — those come later).

```python
removed_features = {
    "Cylinders": "Redundant with Engine Size(L) (r=0.93, VIF~46 together); Engine Size(L) has the "
                 "higher standalone target correlation and is more directly interpretable.",
    "Engine_Size_Per_Cylinder": "Weaker standalone target correlation (r=0.59) than Engine Size(L) "
                                 "alone, and still leaves elevated VIF (~11) when paired with it.",
    "Fuel Consumption City (L/100 km)": "r=0.994 with Comb (L/100 km) -- Comb already captures "
                                         "essentially all of this column's information.",
    "Fuel Consumption Hwy (L/100 km)": "r=0.951 with Comb (L/100 km) -- Comb already captures "
                                        "essentially all of this column's information.",
    "Fuel Consumption Comb (mpg)": "A reciprocal unit-conversion of Comb (L/100 km); carries no "
                                    "information beyond it (VIF in the tens of thousands together).",
    "Transmission (raw code)": "Gear-count suffix showed no consistent within-family relationship "
                                "with CO2 and is undefined for CVTs; replaced by Transmission_Family.",
    "Model": "Extremely high cardinality (2,048 categories) relative to dataset size flagged a "
             "dimensionality concern during EDA. Retained here because target encoding (Stage 8) "
             "collapses it to a single numeric column, resolving the cardinality problem directly "
             "-- see Stage 9 for the residual leakage/multicollinearity trade-off this introduces.",
}

for feature, reason in removed_features.items():
    marker = "KEPT (see note)" if feature == "Model" else "REMOVED"
    print(f"[{marker:15s}] {feature}\n    -> {reason}\n")
```

**Interpretation:** even after removing every mathematically-redundant column, `Engine Size(L)` and

residual collinearity cannot be engineered away without discarding real information from one of the
two columns; we carry both forward and revisit this explicitly, with actual trade-offs, in Stage 9
once the full encoded feature matrix is available.

```python
df_engineered = df_clean[final_features + [target_col]].copy()
print(f"Engineered dataset shape: {df_engineered.shape}")
df_engineered.head()
```

# Stage 7: Train-Test Split

We split **before** fitting any encoder, so that every subsequent preprocessing step (Stage 8) can
be fit exclusively on the training portion and applied to the test portion without leaking
information about test-set categories or distributions into the training process.

```python
X = df_engineered[final_features].copy()
y = df_engineered[target_col].copy()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print(f"X_train: {X_train.shape}   y_train: {y_train.shape}")
print(f"X_test : {X_test.shape}   y_test : {y_test.shape}")
```

**Conclusion:** `X_train`/`y_train` (80%) and `X_test`/`y_test` (20%) are now fixed for the rest of
this notebook. No statistic derived from `X_test`/`y_test` is used to fit anything below.

# Stage 8: Feature Encoding

**Rule for this entire stage: every encoder is fit on `X_train`/`y_train` only, then applied
(transformed) onto `X_test`.** No encoder ever sees the test set during fitting — that would leak
information about test-set categories (or, for target encoding, test-set target values) back into
the features used to predict them.

## 8.1 Target Encoding — `Make` and `Model`

`Make` (42 categories) and `Model` (2,048 categories) are both high-cardinality categorical
features. One-hot encoding `Model` alone would add ~2,000 mostly-empty columns — a sparse,
memory-heavy representation that would overwhelm the six other features and make linear-model
coefficients nearly uninterpretable.

**Why target encoding is appropriate here:** it replaces each category with a single number (the
training-set mean target value for that category), collapsing an arbitrarily-large categorical
column into exactly one numeric column, while still preserving the category's relationship to the
outcome — unlike one-hot encoding, dimensionality does not grow with the number of categories.

**Why fitting only on the training set prevents leakage:** the encoding value for a category is
computed from `y_train` values only. If we computed it from the full dataset (train + test), each
test row's encoded feature would be partly derived from that same row's own target value —
information the model is never supposed to have access to at prediction time. Fitting on
`X_train`/`y_train` alone and only *mapping* those learned means onto `X_test` avoids this exactly.

```python
make_category_wise_mean = y_train.groupby(X_train["Make"].astype(object)).mean()
model_category_wise_mean = y_train.groupby(X_train["Model"].astype(object)).mean()
global_mean = y_train.mean()

print("Sample of learned Make category means (training data only):")
print(make_category_wise_mean.head())
print(f"\nGlobal training mean (fallback value): {global_mean:.2f}")
```

```python
unseen_makes_in_test = set(X_test["Make"].unique()) - set(X_train["Make"].unique())
unseen_models_in_test = set(X_test["Model"].unique()) - set(X_train["Model"].unique())

print(f"Unseen Make categories in test set : {len(unseen_makes_in_test)}")
print(f"Unseen Model categories in test set: {len(unseen_models_in_test)} "
      f"out of {X_test['Model'].nunique()} unique test-set models")
```

**Interpretation:** every `Make` in the test set was already observed during training (0 unseen),
but roughly **18% of the test set's `Model` values were never seen during training** — a direct
consequence of `Model` having 2,048 categories spread across only ~6,300 rows. Any unseen category
must fall back to a sensible default; we use the overall training-set mean target value as that
fallback, so an unfamiliar model is encoded as "an average vehicle" rather than causing an error or
a biased guess.

**A caveat worth stating plainly:** because so many `Model` categories have only 1-2 training rows
(roughly 40% of training-set models appear exactly once), the target-encoded value for those rows is
nearly equal to that row's own actual CO2 value. This is a known weakness of plain mean target
encoding on very high-cardinality, low-frequency categories — it behaves almost like memorization
for rare categories, and its effect on multicollinearity is examined concretely in Stage 9.

### Productionizing the Encoding — a Reusable, Pipeline-Compatible Transformer

The manual `groupby().mean()` logic above is correct but ad hoc. We wrap the identical logic in a
scikit-learn-compatible transformer so it can be composed inside a `ColumnTransformer`/`Pipeline`
alongside one-hot encoding, guaranteeing the same train-only-fit discipline is enforced structurally
rather than by convention.

```python
class MeanTargetEncoder(BaseEstimator, TransformerMixin):
    """Encodes each category as its training-set mean target value.

    Unseen categories at transform time fall back to the overall training-set mean,
    matching the manual approach demonstrated above.
    """

    def __init__(self, columns):
        self.columns = columns

    def fit(self, X, y):
        y = pd.Series(np.asarray(y), index=X.index)
        self.category_means_ = {
            col: y.groupby(X[col].astype(object)).mean() for col in self.columns
        }
        self.global_mean_ = float(y.mean())
        return self

    def transform(self, X):
        encoded = pd.DataFrame(index=X.index)
        for col in self.columns:
            encoded[col] = (
                X[col].astype(object).map(self.category_means_[col])
                .fillna(self.global_mean_)
                .astype(float)
            )
        return encoded.to_numpy()

    def get_feature_names_out(self, input_features=None):
        return np.array([f"{col}_target_enc" for col in self.columns])
```

## 8.2 One-Hot Encoding — `Vehicle Class`, `Fuel Type` (and `Transmission_Family`)

`Vehicle Class` (16 categories) and `Fuel Type` (5 categories) are low-cardinality with no natural
order, so one-hot encoding is appropriate; `drop='first'` avoids the dummy-variable trap (perfect
multicollinearity between a full set of dummies and an intercept). `Transmission_Family` (5
categories, created in Stage 5) was not named explicitly in the encoding requirements, but is
encoded the same way here for the same reason — it is a low-cardinality nominal feature with no
ordinal structure.

```python
one_hot_columns = ["Vehicle Class", "Fuel Type", "Transmission_Family"]
target_encode_columns = ["Make", "Model"]

preprocessor = ColumnTransformer(
    transformers=[
        ("target_enc", MeanTargetEncoder(columns=target_encode_columns), target_encode_columns),
        ("one_hot", OneHotEncoder(drop="first", handle_unknown="ignore", sparse_output=False),
         one_hot_columns),
    ],
    remainder="passthrough",
    verbose_feature_names_out=False,
)

preprocessor.fit(X_train, y_train)

X_train_encoded = pd.DataFrame(
    preprocessor.transform(X_train),
    columns=preprocessor.get_feature_names_out(),
    index=X_train.index,
).astype(float)

X_test_encoded = pd.DataFrame(
    preprocessor.transform(X_test),
    columns=preprocessor.get_feature_names_out(),
    index=X_test.index,
).astype(float)

print(f"X_train_encoded shape: {X_train_encoded.shape}")
print(f"X_test_encoded shape : {X_test_encoded.shape}")
X_train_encoded.head()
```

**Interpretation:** the `ColumnTransformer` was fit once, on `X_train` and `y_train` only, and that
single fitted object produced both `X_train_encoded` and `X_test_encoded` via `.transform()` — the
test set never influenced the learned category means or the one-hot category vocabulary. The
7 original selected features expand to **26 encoded columns**: 2 target-encoded numeric columns
(`Make_target_enc`, `Model_target_enc`), 23 one-hot dummy columns across the three nominal features
(after dropping one reference category each), and the 2 numerical features passed through
unchanged (`Engine Size(L)`, `Fuel Consumption Comb (L/100 km)`).

# Stage 9: Multicollinearity Analysis (Post-Encoding)

Encoding can introduce *new* multicollinearity that wasn't visible in the raw data — most notably,
target encoding directly injects target-related information into a feature, which can interact with
other strong predictors in ways plain correlation on raw categories could never reveal. We recompute
VIF on the fully encoded **training** set only (consistent with "fit on train only" throughout).

```python
vif_encoded = pd.DataFrame({
    "feature": X_train_encoded.columns,
    "VIF": [
        variance_inflation_factor(X_train_encoded.values, i)
        for i in range(X_train_encoded.shape[1])
    ],
}).sort_values("VIF", ascending=False).reset_index(drop=True)

vif_encoded
```

```python
high_vif = vif_encoded[vif_encoded["VIF"] > 10]
print(f"Features with VIF > 10: {len(high_vif)} of {len(vif_encoded)}")
high_vif
```

**Interpretation:**

- **`Fuel Consumption Comb (L/100 km)` (VIF ~509) and `Model_target_enc` (VIF ~448) are now the
  most extreme values in the entire notebook** — far higher than anything seen on the raw numerical
  columns. This is the concrete manifestation of the caveat raised in Stage 8: because ~40% of
  `Model` categories have only one training row, `Model_target_enc` is, for those rows, almost
  identical to the row's own CO2 value — which makes it (and by extension, anything correlated with
  emissions, like fuel consumption) appear almost perfectly predictable from the rest of the matrix.
  This is a **target-encoding-induced leakage artifact**, not a genuine relationship between
  physical vehicle attributes.
- **`Make_target_enc` (VIF ~61)** shows the same effect at a smaller scale — `Make` has far larger,
  more stable groups (42 categories, ~120 rows each on average) than `Model`, so the effect is real
  but much less severe.
- **`Engine Size(L)` (VIF ~30)** is elevated beyond its pre-encoding pairwise value (~17) because it
  is now also partially explainable through the target-encoded columns, which themselves correlate
  with engine size (larger engines -> higher emissions -> higher target-encoded value for that
  vehicle's make/model).
- **`Fuel Type_Z` and `Fuel Type_X` (VIF ~18-20)** are moderately elevated, but this is a normal,
  expected side effect of one-hot encoding two dominant categories that together cover the large
  majority of the fleet — not a cause for concern on its own.
- **The remaining one-hot dummy columns (`Vehicle Class_*`, `Transmission_Family_*`) all show
  VIF well under 5** — these encode cleanly with no meaningful redundancy.

**Recommendation (not applied automatically — a trade-off, not a rule):**

| Option | Interpretability | Information Loss | Prediction Accuracy Risk |
|---|---|---|---|
| Replace plain mean target encoding on `Model` with a regularized/smoothed variant (e.g. shrinkage toward the `Make` or global mean, or out-of-fold encoding) | Slightly lower (encoded value is a blend, not a raw mean) | Low — most of the model-specific signal is retained, just dampened for rare categories | Reduces overfitting risk from near-memorization on singleton categories |
| Drop `Model_target_enc` entirely, keep only `Make_target_enc` | Highest (fewer, more stable features) | Moderate — loses model-specific nuance beyond brand-level averages | Likely more stable out-of-sample, since brand-level groups are far less prone to leakage |
| Drop one of `Engine Size(L)` / `Fuel Consumption Comb (L/100 km)` | Trade interpretability (engine displacement is policy-relevant) against a purer, more direct emissions driver (fuel consumption) | Moderate either way — the two are correlated (r ~ 0.82) but not identical | Regularized linear models (Ridge) tolerate this residual collinearity reasonably well without dropping either |

No features are removed in this notebook. These trade-offs are surfaced here so the modeling
notebook can make an informed, model-family-specific choice (e.g. Ridge/Lasso regularization can
absorb much of this residual collinearity without manual removal).

# Preprocessing Summary

**Feature engineering decisions**

- `Transmission` -> `Transmission_Family`: gear-count suffix dropped (no consistent within-family
  relationship with CO2; undefined for CVTs); family retained (5 categories).
- Engine representation: `Engine Size(L)` retained; `Cylinders` and the engineered
  `Engine_Size_Per_Cylinder` ratio dropped (redundant, weaker, or both).
- Fuel consumption: `Fuel Consumption Comb (L/100 km)` retained; `City`, `Hwy`, and `Comb (mpg)`
  dropped as mathematically/statistically redundant.

**Final selected features (7):** `Engine Size(L)`, `Fuel Consumption Comb (L/100 km)`, `Make`,
`Model`, `Vehicle Class`, `Transmission_Family`, `Fuel Type`.

**Train/test split:** 80/20, `random_state=42`, performed before any encoder was fit.

**Encoding:** `Make` and `Model` -> training-set mean target encoding (via a reusable
`MeanTargetEncoder`, fit on `X_train`/`y_train` only); `Vehicle Class`, `Fuel Type`,
`Transmission_Family` -> one-hot encoding with `drop='first'` (fit on `X_train` only). Both
combined in a single `ColumnTransformer`, producing 26 final encoded columns.

**Post-encoding multicollinearity:** severe residual VIF on `Fuel Consumption Comb (L/100 km)`,
`Model_target_enc`, and `Make_target_enc`, traced to target-encoding leakage on very low-frequency
`Model` categories, plus a genuine physical Engine-Size/Fuel-Consumption relationship. No features
were removed; trade-offs are documented above for the modeling stage to resolve.

**Deliverables produced by this notebook:** `df_engineered` (final engineered dataset),
`final_features` (final selected feature list), `X_train_encoded` / `X_test_encoded` (encoded
train/test feature matrices), `y_train` / `y_test`, and `vif_encoded` (final VIF table).

Scaling and model training are explicitly out of scope here and will be addressed in the next
notebook.

# Stage 10: Model B — Interpretable Policy Model

Stages 1-9 above deliberately stopped short of scaling and model training. Following a design
review against the three source documents (business questions, approach document, dataset
description), the priority for this project is **understanding which vehicle design and policy
levers influence CO2 emissions**, not squeezing out the last percentage point of predictive
accuracy. That review produced "Model B": a deliberately narrower, interpretable regression rather
than a model using every available feature.

**Model B's feature set:** `Engine Size(L)`, `Make`, `Vehicle Class`, `Transmission_Family`,
`Fuel Type`.

**Deliberately excluded, and why:**
- **`Fuel Consumption Comb (L/100 km)`** — CO2 output from a combustion engine is chemically almost
  a fixed multiple of fuel burned. Fuel consumption therefore sits *between* the vehicle's design
  attributes and its emissions (Engine Size/Vehicle Class -> Fuel Consumption -> CO2). Including it
  would let the model "explain" emissions via combustion chemistry rather than via the structural,
  actionable vehicle attributes the Global Automotive Council can actually influence — and it would
  crowd out Engine Size's and Vehicle Class's coefficients in the process (classic mediator-inclusion
  bias).
- **`Model`** — with 2,048 categories over ~6,300 rows, this behaves as a near-identifier for a
  specific trim rather than a business-actionable lever (you cannot regulate "the LT trim vs. the
  LS trim"). It is also largely redundant with Make + Engine Size + Vehicle Class + Fuel Type once
  those are known.

We reuse `X_train`/`X_test`/`y_train`/`y_test` from Stage 7 — same 80/20 split, same
`random_state=42` — simply selecting a narrower column subset. No re-cleaning or re-splitting is
needed.

### Additional Imports for Model B

Stages 5-9 needed `train_test_split`, `OneHotEncoder`, and `ColumnTransformer`. Model B additionally
needs feature scaling and a regression estimator with built-in cross-validated regularization
selection, plus standard regression evaluation metrics.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import RidgeCV
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
model_b_features = ["Engine Size(L)", "Make", "Vehicle Class", "Transmission_Family", "Fuel Type"]

X_train_b = X_train[model_b_features].copy()
X_test_b = X_test[model_b_features].copy()

print(f"X_train_b: {X_train_b.shape}   X_test_b: {X_test_b.shape}")
X_train_b.head()
```

# Stage 11: Encoding and Scaling for Model B

Ridge regression penalizes coefficient magnitude directly, so it is sensitive to the *scale* of each
feature — a feature measured in large raw units (Engine Size, roughly 0.9-8.4) would otherwise be
penalized unfairly relative to already-bounded 0/1 dummy columns. We therefore standardize
`Engine Size(L)` (zero mean, unit variance) but **deliberately leave the one-hot dummies as raw
0/1** rather than standardizing them too: a dummy coefficient then reads directly as "g/km CO2 versus
the baseline category," the most business-readable unit available, whereas standardizing a binary
column would turn its coefficient into an awkward "per-SD-of-a-binary" quantity with no natural
interpretation.

`Make` (42 categories) now uses **one-hot encoding**, replacing the Target Encoding used in the
original Stage 8 pipeline — with only 42 categories this is a perfectly manageable one-hot size, and
one-hot gives an individually auditable coefficient per brand instead of one opaque blended scalar,
directly serving this model's interpretability goal. `Vehicle Class`, `Transmission_Family`, and
`Fuel Type` continue to use one-hot encoding as before. As throughout this notebook, every encoder
is fit on `X_train_b`/`y_train` only.

```python
categorical_features_b = ["Make", "Vehicle Class", "Transmission_Family", "Fuel Type"]

preprocessor_b = ColumnTransformer(
    transformers=[
        ("scale", StandardScaler(), ["Engine Size(L)"]),
        ("onehot", OneHotEncoder(drop="first", handle_unknown="ignore", sparse_output=False),
         categorical_features_b),
    ],
    remainder="drop",
    verbose_feature_names_out=False,
)

preprocessor_b.fit(X_train_b, y_train)

X_train_b_encoded = pd.DataFrame(
    preprocessor_b.transform(X_train_b),
    columns=preprocessor_b.get_feature_names_out(),
    index=X_train_b.index,
).astype(float)

X_test_b_encoded = pd.DataFrame(
    preprocessor_b.transform(X_test_b),
    columns=preprocessor_b.get_feature_names_out(),
    index=X_test_b.index,
).astype(float)

print(f"X_train_b_encoded: {X_train_b_encoded.shape}")
print(f"X_test_b_encoded : {X_test_b_encoded.shape}")
print()
print("Baseline (reference) categories dropped by one-hot encoding:")
for col in categorical_features_b:
    baseline = sorted(X_train_b[col].unique())[0]
    print(f"  {col:20s} baseline = {baseline}")
```

**Interpretation:** the fitted `ColumnTransformer` expands 5 raw features into a fully numeric
encoded matrix. Every dummy coefficient computed in Stage 14 will be relative to one of the printed
baseline categories (e.g. `Vehicle Class` coefficients are read relative to `COMPACT`, `Fuel Type`
relative to Diesel, `Make` relative to Acura, `Transmission_Family` relative to conventional
Automatic). This is standard practice for `drop='first'` one-hot encoding and simply needs to be
kept in mind when narrating coefficients later.

# Stage 12: Train Ridge Regression (Cross-Validated Regularization Strength)

Ridge regression is the right tool here per the earlier review: the remaining features still carry
some genuine, non-redundant correlation (e.g. larger engines tend to appear in larger vehicle
classes), and Ridge shrinks correlated coefficients together rather than letting OLS distribute
weight between them erratically, without forcing an arbitrary "drop one of them" decision that OLS's
instability would otherwise invite.

Rather than picking a regularization strength by hand, we search over a wide grid of `alpha` values
with 5-fold cross-validation on the training set only, per the Approach document's guidance to use
cross-validation so the model doesn't overfit to any particular subset of the data.

```python
alpha_grid = np.logspace(-3, 3, 50)
ridge_model = RidgeCV(alphas=alpha_grid, cv=5)
ridge_model.fit(X_train_b_encoded, y_train)

print(f"Alpha grid searched : [{alpha_grid.min():.4f}, {alpha_grid.max():.1f}] ({len(alpha_grid)} values)")
print(f"Selected alpha      : {ridge_model.alpha_:.4f}")
```

**Interpretation:** the selected alpha (~0.12) sits comfortably inside the searched grid rather than
at either edge, confirming the grid was wide enough and the choice is a genuine interior optimum
rather than an artifact of a too-narrow search range. A small alpha indicates the training data does
not need heavy shrinkage to generalize — consistent with the moderate (not severe) residual
collinearity left in this feature set.

# Stage 13: Evaluate Model B

Per the Approach document's guidance to report several complementary metrics, we use three: **R²**
(share of variance in CO2 emissions explained), **RMSE** (average error, penalizing large misses more
heavily), and **MAE** (typical/average-case error, directly in g/km — the most business-readable of
the three). We report both training and test performance side by side to check for overfitting.

```python
def regression_report(y_true, y_pred, label):
    r2 = r2_score(y_true, y_pred)
    rmse = mean_squared_error(y_true, y_pred) ** 0.5
    mae = mean_absolute_error(y_true, y_pred)
    print(f"{label:6s} ->  R2 = {r2:.4f}   RMSE = {rmse:6.2f} g/km   MAE = {mae:6.2f} g/km")
    return r2, rmse, mae


y_pred_train = ridge_model.predict(X_train_b_encoded)
y_pred_test = ridge_model.predict(X_test_b_encoded)

regression_report(y_train, y_pred_train, "Train")
test_r2, test_rmse, test_mae = regression_report(y_test, y_pred_test, "Test")

print(f"\nTest MAE as % of average test-set emissions: {test_mae / y_test.mean() * 100:.1f}%")
```

```python
fig, ax = plt.subplots(figsize=(7, 7))
ax.scatter(y_test, y_pred_test, alpha=0.3, s=15, color="#3b6ea5")
lims = [min(y_test.min(), y_pred_test.min()), max(y_test.max(), y_pred_test.max())]
ax.plot(lims, lims, color="#c0392b", linewidth=2, label="Perfect prediction (y = x)")
ax.set_xlabel("Actual CO2 Emissions (g/km)")
ax.set_ylabel("Predicted CO2 Emissions (g/km)")
ax.set_title("Model B — Predicted vs. Actual CO2 Emissions (Test Set)")
ax.legend()
fig.tight_layout()
plt.show()
```

**Interpretation:** Model B explains roughly **87% of the variance** in CO2 emissions (test R² ~
0.87) using only Engine Size, Make, Vehicle Class, Transmission Family, and Fuel Type — with **no**
fuel-consumption or model-identity information at all. The typical prediction error (MAE) is about
16 g/km, roughly **6% of the average emissions level** in the test set — a small error relative to
the scale of the target. Training and test R² are nearly identical (~0.873 vs ~0.872), indicating
the model generalizes well and is not overfitting.

The predicted-vs-actual scatter clusters tightly around the reference line across the bulk of the
range, with the visible scatter widening somewhat at the high-emission end — consistent with the
worst-residual vehicles examined in Stage 15 being concentrated among a small number of extreme
sports/performance cars and hybrids. There is no obvious systematic curve (e.g. consistent
over- or under-prediction across the whole range), which would have signaled a poorly-specified
functional form; the errors look like localized exceptions rather than a global fit problem.

# Stage 14: Interpret Ridge Coefficients — Which Factors Matter Most?

This is the core deliverable of Model B. Every coefficient below represents the estimated change in
CO2 emissions (g/km) associated with that feature, holding all other included features constant —
`Engine Size(L)`'s coefficient is per one standard deviation of engine size, and every categorical
coefficient is relative to its baseline category printed in Stage 11.

```python
ridge_coefficients = (
    pd.Series(ridge_model.coef_, index=X_train_b_encoded.columns, name="coefficient")
    .sort_values(key=abs, ascending=False)
)

print("Top 15 features by |coefficient| (largest estimated effect on CO2 emissions):")
ridge_coefficients.head(15).to_frame()
```

```python
def plot_top_coefficients(coefficients, n=15):
    top = coefficients.head(n).sort_values()
    colors = ["#c0392b" if v > 0 else "#3b6ea5" for v in top.values]

    fig, ax = plt.subplots(figsize=(9, 8))
    ax.barh(top.index, top.values, color=colors)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlabel("Ridge coefficient (g/km CO2, vs. baseline category)")
    ax.set_title(f"Model B — Top {n} Drivers of CO2 Emissions by |Coefficient|")
    fig.tight_layout()
    plt.show()


plot_top_coefficients(ridge_coefficients, n=15)
```

**Interpretation:**

- **`Engine Size(L)` (+41 g/km per standard deviation)** is, as expected from EDA, a strong,
  business-actionable driver — this is the clearest "engine downsizing reduces emissions" signal in
  the model, and it survives being estimated *alongside* Make and Vehicle Class rather than in
  isolation, which EDA's raw correlation alone could not confirm.
- **Vehicle Class dominates the top of the list**: `VAN - PASSENGER` (+115) and `VAN - CARGO` (+83)
  add the most CO2 versus the `COMPACT` baseline, with `PICKUP TRUCK` and `SUV - STANDARD` classes
  also elevated (+34 to +55). This is a direct, actionable "fleet composition" insight: shifting
  fleet mix away from vans/large SUVs/trucks and toward compact-class vehicles is the single largest
  lever visible in this model, independent of engine size or brand.
- **Exotic/performance brands show the largest positive coefficients of all** (`BUGATTI` +131,
  `LAMBORGHINI` +92, `MASERATI` +78, `BENTLEY` +74, `ROLLS-ROYCE` +58, `ASTON MARTIN` +56) — these
  brands add substantial CO2 even after controlling for engine size and vehicle class, suggesting
  brand captures real engineering/tuning philosophy differences (e.g. performance-tuned engines)
  beyond what displacement alone explains.
- **A caution on brand coefficients:** these coefficients are only as reliable as their sample size.
  `BUGATTI`'s coefficient is estimated from a **single training-set vehicle**; `GENESIS` from 11.
  Ridge's shrinkage tempers this somewhat, but treat the Bugatti figure as anecdotal rather than a
  generalizable brand effect, while the broader luxury/performance cluster (Lamborghini, Bentley,
  Rolls-Royce, Aston Martin — each with 26-40 training examples) is on firmer statistical footing and
  tells a consistent, credible story.
- **Cross-check against EDA:** EDA's raw group-means (Section 5.C.2) showed similar qualitative
  patterns (vans/trucks and exotic brands highest, compacts/station wagons lowest) — the Ridge
  coefficients broadly agree with, and sharpen, that earlier univariate picture, which is reassuring:
  the "brand and vehicle-class effects" story is not simply an artifact of larger-engined vehicles
  happening to cluster in those categories, since Engine Size is already held constant here.

# Stage 15: Residual Analysis — Which Vehicles Does the Model Get Wrong?

Beyond aggregate metrics, inspecting the largest individual errors tells us *where* the model's
assumptions break down — directly answering the case-study question about vehicles with unusually
high or low emissions relative to similar peers.

```python
test_residuals = y_test - y_pred_test

residual_table = pd.DataFrame({
    "actual": y_test,
    "predicted": np.round(y_pred_test, 1),
    "residual": np.round(test_residuals, 1),
    "abs_error": np.round(test_residuals.abs(), 1),
}).sort_values("abs_error", ascending=False)

worst_cases = residual_table.head(12).join(
    df_clean.loc[residual_table.head(12).index, ["Make", "Model", "Vehicle Class", "Fuel Type"]]
)
worst_cases[["Make", "Model", "Vehicle Class", "Fuel Type", "actual", "predicted", "residual"]]
```

**Interpretation — two distinct failure patterns emerge, both explainable by what Model B
deliberately excludes:**

- **Hybrid vehicles are consistently over-predicted** (Hyundai Ioniq, Kia Niro FE, Kia Optima
  Hybrid, Volkswagen Jetta Turbo Hybrid all show large *negative* residuals — actual emissions far
  below prediction). `Fuel Type` in this dataset does not distinguish hybrid drivetrains from
  conventional ones, so the model has no way to know these vehicles achieve unusually low emissions
  for their engine size/class. This is a genuine, actionable finding: **hybrid/electrification status
  is a missing signal**, not a modeling flaw — if hybrid-drivetrain flags were available, they would
  likely be a strong additional predictor.
- **Extreme sports/performance cars are consistently under-predicted** (Ford GT, Audi R8 Spyder,
  Chevrolet Corvette, Lamborghini Aventador — all large *positive* residuals). These are vehicles
  whose performance tuning pushes emissions well above what their brand/class/engine-size profile
  would typically suggest — exactly the "high-performance vehicles as genuine outliers" pattern the
  Approach document anticipated. This reinforces the Stage 5.F decision not to remove such outliers:
  they are real vehicles behaving in a genuinely different regime, not data errors.
- **Heavy off-road SUVs are also under-predicted** (Mercedes-Benz G 550, AMG G 63) — likely reflecting
  vehicle weight/off-road drivetrain characteristics that are correlated with, but not fully captured
  by, `Vehicle Class` and `Engine Size` alone.

None of this suggests Model B is broken — it suggests the model's simplicity (by design) leaves a
specific, identifiable, and explainable residual pattern, which is itself a useful business insight
about the limits of what "vehicle design category" alone can predict.

# Stage 16: Business Insights & Recommendations

**Which factors most strongly influence CO2 emissions?** In order of estimated impact, holding other
factors constant: (1) **vehicle class**, especially vans, pickup trucks, and standard SUVs, versus
compact-class vehicles; (2) **brand/manufacturer**, with exotic/performance brands showing a
substantial premium even after controlling for engine size and class; (3) **engine size**, a
consistent, moderate, and directly actionable driver.

**How this supports emission-reduction efforts:**

- **Fleet composition is the single largest lever identified.** Policies that shift vehicle mix away
  from large SUVs/vans/trucks toward compact-class vehicles would plausibly have the largest
  aggregate emissions impact of any lever examined here — larger than engine-size regulation alone.
- **Engine downsizing remains a valid, complementary lever** — Model B confirms Engine Size's effect
  is real and additive to the vehicle-class effect, not merely a proxy for it.
- **Brand-level differences suggest engineering/tuning philosophy matters independent of size and
  class** — this is useful context for benchmarking manufacturers against their size-and-class peers
  rather than against the market as a whole, though the smallest-volume luxury brands' coefficients
  should be treated as indicative rather than precise (Stage 14 caution).
- **Hybrid/electrified drivetrains are conspicuously absent from this dataset's feature set, and the
  residual analysis (Stage 15) shows exactly why that matters**: hybrids in this data achieve
  emissions far below what their engine size and class would predict. If the Council's goal is
  emissions reduction, incentivizing hybrid/electrified powertrains within any vehicle class or
  engine-size segment is supported by direct evidence in this dataset, even though "hybrid" is not
  itself a modeled feature here.
- **Extreme-performance vehicles are a small-volume, high-impact-per-vehicle segment** — regulation
  targeted specifically at high-performance trims (rather than uniform per-class rules) could
  address a disproportionate share of emissions from a small number of vehicles.
